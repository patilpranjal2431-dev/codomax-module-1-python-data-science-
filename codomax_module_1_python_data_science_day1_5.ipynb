```python
# 1. Student Grade Calculator
def calculate_grade(marks):
    if 90 <= marks <= 100:
        return "A"
    elif 75 <= marks <= 89:
        return "B"
    elif 60 <= marks <= 74:
        return "C"
    elif 40 <= marks <= 59:
        return "D"
    elif 0 <= marks < 40:
        return "F"
    else:
        return "Invalid Marks"

test_marks = [95, 82, 67, 50, 28]

for mark in test_marks:
    print(f"Marks: {mark} -> Grade: {calculate_grade(mark)}")
```

```python
# 2. Count Vowels
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for char in text:
        if char in vowels:
            count += 1

    return count

# Test the function
input_text = "Python Programming"
result = count_vowels(input_text)

print(f"Input: \"{input_text}\"")
print(f"Output: {result}")
```

```python
# 3. Calculate Factorial
def factorial(number):
    if number < 0:
        return "Factorial does not exist for negative numbers."
    elif number == 0 or number == 1:
        return 1

    result = 1
    for i in range(2, number + 1):
        result *= i
    return result

test_numbers = [5, 0, 1, 7, -3]

for num in test_numbers:
    print(f"factorial({num}) -> {factorial(num)}")
```

```python
# 4. Student Performance Analyzer
def analyze_marks(marks):
    if not marks:
        return "The list of marks is empty."

    total_marks = sum(marks)
    average_marks = total_marks / len(marks)
    highest_marks = max(marks)
    lowest_marks = min(marks)

    passed_students = sum(1 for mark in marks if mark >= 40)
    failed_students = len(marks) - passed_students

    return {
        "Average Marks": round(average_marks, 2),
        "Highest Marks": highest_marks,
        "Lowest Marks": lowest_marks,
        "Passed Students": passed_students,
        "Failed Students": failed_students
    }

student_marks = [85, 32, 90, 78, 25, 60, 45, 92, 38]
analysis = analyze_marks(student_marks)

print("--- Student Performance Analysis ---")
for key, value in analysis.items():
    print(f"{key}: {value}")
```

```python
# 10. Mini Employee Salary Analyzer

def calculate_salary(basic_salary, bonus_percentage, tax_percentage):
    bonus = basic_salary * bonus_percentage / 100
    gross_salary = basic_salary + bonus
    tax = gross_salary * tax_percentage / 100
    net_salary = gross_salary - tax
    return net_salary

def salary_category(net_salary):
    if net_salary < 30000:
        return "Low"
    elif 30000 <= net_salary <= 60000:
        return "Medium"
    else:
        return "High"

employees = [
    {"name": "Alice", "basic": 25000, "bonus_pct": 10, "tax_pct": 10},
    {"name": "Bob", "basic": 45000, "bonus_pct": 15, "tax_pct": 12},
    {"name": "Charlie", "basic": 75000, "bonus_pct": 20, "tax_pct": 15}
]

print("--- Employee Salary Analysis ---")
for emp in employees:
    net = calculate_salary(emp["basic"], emp["bonus_pct"], emp["tax_pct"])
    category = salary_category(net)
    print(f"{emp['name']} -> Net Salary: ₹{net:,.2f}  Category: {category}")
```

```python
# 2. Find Statistics from a List
numbers = [12, 45, 23, 67, 34, 89, 10, 56]

count = len(numbers)
total_sum = sum(numbers)
average = total_sum / count

largest = numbers[0]
smallest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num

print("Largest number:", largest)
print("Smallest number:", smallest)
print("Sum:", total_sum)
print("Average:", average)
print("Number of elements:", count)
```

```python
# 3. Separate Even and Odd Numbers
numbers = [12, 7, 25, 18, 30, 41, 56, 63, 72, 9]

even_numbers = []
odd_numbers = []

for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)

print("Even:", even_numbers)
print("Odd:", odd_numbers)
```

```python
# 7. Mini Employee Data Analyzer

employees = [
    {"name": "Aisha", "department": "IT", "salary": 50000},
    {"name": "Rahul", "department": "HR", "salary": 45000},
    {"name": "Sneha", "department": "IT", "salary": 65000},
    {"name": "Arjun", "department": "Finance", "salary": 55000},
    {"name": "Neha", "department": "IT", "salary": 70000}
]

print("--- All Employees ---")
for emp in employees:
    print(f"Name: {emp['name']} | Dept: {emp['department']} | Salary: ₹{emp['salary']:,}")

highest_emp = max(employees, key=lambda x: x["salary"])
print(f"\nHighest Salary: {highest_emp['name']} (₹{highest_emp['salary']:,})")

lowest_emp = min(employees, key=lambda x: x["salary"])
print(f"Lowest Salary: {lowest_emp['name']} (₹{lowest_emp['salary']:,})")

total_salary = sum(emp["salary"] for emp in employees)
average_salary = total_salary / len(employees)
print(f"Average Salary: ₹{average_salary:,.2f}")

print("\n--- IT Department Employees ---")
it_employees = [emp for emp in employees if emp["department"] == "IT"]
for emp in it_employees:
    print(f"Name: {emp['name']} | Salary: ₹{emp['salary']:,}")

high_earners_count = sum(1 for emp in employees if emp["salary"] > 50000)
print(f"\nNumber of employees earning > ₹50,000: {high_earners_count}")
```

```python
# 5. Mini File-Based Student Management System

def add_student():
    print("\n--- Add Student ---")
    name = input("Enter Name: ").strip()
    age = input("Enter Age: ").strip()
    branch = input("Enter Branch: ").strip()
    marks = input("Enter Marks: ").strip()

    with open("students.txt", "a") as file:
        file.write(f"{name},{age},{branch},{marks}\n")

    print(f"Student '{name}' added successfully!\n")


def view_students():
    print("\n--- All Students ---")
    try:
        with open("students.txt", "r") as file:
            lines = file.readlines()

            if not lines:
                print("No student records found.\n")
                return

            for line in lines:
                data = line.strip().split(",")
                if len(data) == 4:
                    name, age, branch, marks = data
                    print(f"Name: {name} | Age: {age} | Branch: {branch} | Marks: {marks}")
            print()
    except FileNotFoundError:
        print("No student records found. (File does not exist yet)\n")


def search_student():
    print("\n--- Search Student ---")
    search_name = input("Enter student name: ").strip().lower()
    found = False

    try:
        with open("students.txt", "r") as file:
            for line in file:
                data = line.strip().split(",")
                if len(data) == 4:
                    name, age, branch, marks = data
                    if name.lower() == search_name:
                        print("\nStudent Found!")
                        print(f"Name: {name}")
                        print(f"Age: {age}")
                        print(f"Branch: {branch}")
                        print(f"Marks: {marks}\n")
                        found = True
                        break

        if not found:
            print(f"No student found with the name '{search_name}'.\n")

    except FileNotFoundError:
        print("No student records found. (File does not exist yet)\n")


def main():
    while True:
        print("=================================")
        print(" STUDENT MANAGEMENT SYSTEM ")
        print("=================================")
        print("1. Add Student")
        print("2. View Students")
        print("3. Search Student")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ").strip()

        if choice == "1":
            add_student()
        elif choice == "2":
            view_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice! Please select between 1 and 4.\n")

if __name__ == "__main__":
    main()
```
